## 📐 Error Metrics

> *How wrong is my model, and in what way?*

---

### 1. MAE: Mean Absolute Error

**On average, how far are my model's predictions away from the actual rent?**

| | |
|---|---|
| ✅ **Why it's useful** | Easy to explain to anyone, easy to interpret |
| ⚠️ **Weakness** | Treats all errors equally a ₹10,000 mistake and a ₹1,000 mistake both contribute proportionally |

---

### 2. MSE: Mean Squared Error

**Same as MAE, but squares the error before averaging.**

| | |
|---|---|
| ❓ **Why squaring?** | Punishes large mistakes to expose them — useful when big mistakes are especially bad, like predicting ₹8,000 for a ₹2,00,000 PG |
| ⚠️ **Weakness** | Not human readable one bad prediction can pull MSE very high |

---

### 3. RMSE: Root Mean Squared Error

**Same as MSE, but applies square root, brings the value back to original unit (actual rupees).**

| | |
|---|---|
| ✅ **Why it's useful** | Still punishes large errors, but is easier to interpret |

---

### ⚖️ RMSE vs MAE  Which to use?

```md
MAE  → A teacher who grades every wrong answer equally, 1 mark off per mistake
RMSE → A teacher who gives exponentially more marks off the more wrong you are
```

## 📊 Eval Metric

> *How good is my model overall?*

---

### R² Score

**Tells us how much (in %) of the rent variation my model can explain.**

Rent price varies across PGs some ₹6,000, some ₹20,000. That variation exists for reasons like location, size, occupancy, etc. R² measures how much of that *"why does rent differ"* your model figured out.

```md
R² = 0.33  →  model explained 33% of why rents differ
             remaining 67%  →  still a mystery to the model
```

> **R² = what % of the story your model figured out. The rest is still untold.**
