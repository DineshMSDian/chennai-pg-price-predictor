````python
# 1.2 Missing values
def missing_info(df):
    missing_info = pd.DataFrame({
        'missing count': df.isna().sum(),
        'missing %': (df.isna().mean() * 100).round(2),
        'Data type': df.dtypes,
    })

    missing_info = missing_info[missing_info['missing count'] > 0]
    missing_info.sort_values('missing %', ascending = False)
    return missing_info

missing_info(df)
````

## 1.2 Missing Values Analysis

### 1.2.1 Score Columns (transit_score, lifestyle_score)

- Both columns are missing **~50%** of values.
- The API simply did not return scores for those listings. The exact reason is unknown the score may be unavailable for that locality or was never computed by NoBroker.
- These are **API-level absences**, not data entry gaps.

**Imputation strategy:**

1. Introduce binary indicator columns to preserve the missingness signal:
   - `transit_score_missing`
   - `lifestyle_score_missing`
2. Impute missing scores using the **median within the same locality** — properties in the same locality share similar accessibility characteristics.
3. If an entire locality has no scores, fill remaining `NaN` values with the **global median**.

---

### 1.2.2 Amenity Columns (`wifi`, `laundry`, and others)

- Amenity columns are missing **~45%** of values.
- Missingness was validated by comparing raw JSON API responses against the live NoBroker listing page for the same properties.

**Validated semantics:**

| Value | Meaning |
|---|---|
| True | PG owner explicitly confirmed the amenity is **available** |
| False | PG owner explicitly confirmed the amenity is **unavailable** |
| NaN | PG owner provided **no information** about this amenity |

**Root cause:** Listings with an empty `amenitiesMap` in the API response provide no amenity information whatsoever. The `NaN` is structural, not accidental.

**Modeling assumption:**
> Missing amenity values are treated as `False` unlisted amenities are assumed unavailable.
> This is a conservative, defensible assumption given the source structure.

---

### 1.2.3 Columns with ~1% Missing Values

- Several columns including the **target variable rent**  have approximately **1% missing values**.
- These rows are **dropped outright**. Losing ~1% of the dataset introduces no meaningful bias and is far safer than imputing the target variable.

---

### 1.2.4 gate_closing_time (~80%+ Missing)

- Over **80%** of values are absent.
- Beyond the missingness, the column carries negligible predictive signal for rent estimation.
- **Decision: drop unconditionally.**

---

### 1.2.5 Decision Summary

| Column Group | Missing % | Decision | Rationale |
|---|---|---|---|
| transit_score, lifestyle_score | ~50% | Retain + locality-median imputation | API-level absence; indicator columns added |
| Amenity columns | ~45% | Impute NaN → False | Validated: missing = owner did not disclose |
| rent + ~1% columns | ~1% | Drop rows | Safe data loss; imputing target is unsound |
| gate_closing_time | ~80%+ | Drop column | High missingness + low predictive value |

## 1.3 Duplicate Analysis

### Observations

- Exact duplicate rows were found in the dataset.
- Likely originated from overlapping API responses or listings that appear under multiple localities (e.g., a PG listed under both *Tambaram* and *Perungalathur*).
- Since all columns are identical across duplicates, deduplication is performed on the natural composite key: **id + occupancy**.

### Decision

Duplicates are removed keeping the first occurrence of each unique property-room combination.

```python
df_clean = df.drop_duplicates(subset=['id', 'occupancy'], keep='first')
```
